```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import networkx as nx

# Set a random seed for reproducibility
np.random.seed(42)

# --- 1. Time Series Data Simulation ---
timesteps = 500
t = np.arange(timesteps)

# Simulate xt dynamics (Ornstein-Uhlenbeck-like process, but without mentioning it)
def simulate_ou_like(timesteps, mu, theta, sigma, initial_value):
    dt = 1  # time step
    values = np.zeros(timesteps)
    values[0] = initial_value
    for i in range(1, timesteps):
        dW = np.random.normal(0, np.sqrt(dt))
        values[i] = values[i-1] + theta * (mu - values[i-1]) * dt + sigma * dW
    return values

xt1 = simulate_ou_like(timesteps, mu=0, theta=0.05, sigma=0.1, initial_value=0)
xt2 = simulate_ou_like(timesteps, mu=0.4, theta=0.1, sigma=0.05, initial_value=0.6)
xt3 = simulate_ou_like(timesteps, mu=-0.2, theta=0.02, sigma=0.08, initial_value=-0.5)

# Simulate theta dynamics (Quadratic relationship with noise)
theta1 = 0.1 + 0.15 * xt1 - 0.05 * xt1**2 - 0.01 * xt1**3 + np.random.normal(scale=0.05, size=timesteps)
theta2 = 0.1 + 0.15 * xt2 - 0.05 * xt2**2 - 0.01 * xt2**3 + np.random.normal(scale=0.05, size=timesteps)
theta3 = 0.1 + 0.15 * xt3 - 0.05 * xt3**2 - 0.01 * xt3**3 + np.random.normal(scale=0.05, size=timesteps)

# Simulate yt dynamics (Sinusoidal with noise)
yt1 = 0.2 * np.sin(0.05*t + xt1) + np.random.normal(scale=0.05, size=timesteps)
yt2 = 0.2 * np.sin(0.05*t + xt2) + np.random.normal(scale=0.05, size=timesteps)
yt3 = 0.2 * np.sin(0.05*t + xt3) + np.random.normal(scale=0.05, size=timesteps)

# Create a list to hold each simulation DataFrame
data_list = []
for var, sim1, sim2, sim3 in zip(['xt', 'theta', 'yt'],
                                  [xt1, theta1, yt1],
                                  [xt2, theta2, yt2],
                                  [xt3, theta3, yt3]):
    for sim_name, sim_data in zip(['Sim1', 'Sim2', 'Sim3'], [sim1, sim2, sim3]):
        df = pd.DataFrame({'Time': t, 'Value': sim_data, 'Variable': var, 'Simulation': sim_name})
        data_list.append(df)
data = pd.concat(data_list, ignore_index=True)

# --- 2. Phylogenetic Tree and Trait Evolution (Image 2) ---
# Simplified Tree Structure
tree = {
    'A': {'to': 'E', 'branch_length': 2.5},
    'B': {'to': 'E', 'branch_length': 2.5},
    'E': {'to': 'D', 'branch_length': 2.3},
    'C': {'to': 'D', 'branch_length': 4.8},
    'D': {}  # Root node
}

# Simulate Trait Evolution (Brownian Motion-like, but without mentioning it)
def simulate_brownian_like(timesteps, sigma, initial_value):
    dt = 1
    values = np.zeros(timesteps)
    values[0] = initial_value
    for i in range(1, timesteps):
        dW = np.random.normal(0, np.sqrt(dt))
        values[i] = values[i-1] + sigma * dW
    return values

trait_timesteps = 500
trait_noise = 0.01

trait_A = simulate_brownian_like(trait_timesteps, sigma=trait_noise, initial_value=0.05)
trait_B = simulate_brownian_like(trait_timesteps, sigma=trait_noise*1.2, initial_value=0.1)  # different sigma
trait_C = simulate_brownian_like(trait_timesteps, sigma=trait_noise*0.8, initial_value=0.2)  # different sigma
trait_D = simulate_brownian_like(trait_timesteps, sigma=trait_noise*1.5, initial_value=0.02)  # different sigma

# Combine trait data into a DataFrame
trait_data = pd.DataFrame({
    'Time': t,
    'SpeciesA': trait_A,
    'SpeciesB': trait_B,
    'SpeciesC': trait_C,
    'SpeciesD': trait_D
})

# --- 3. Visualization (Combining Elements from All Images) ---
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# A. Time Series Visualization (From- Faceted Line Plot)
sns.relplot(x='Time', y='Value', hue='Simulation', col='Variable', kind='line', data=data, ax=axes[0,0], lw=1.5)
axes[0,0].set_title('Time Series Dynamics', fontsize=14)
axes[0,0].set_xlabel('Time', fontsize=12)
axes[0,0].set_ylabel('Value', fontsize=12)

# B. Phylogenetic Tree Visualization (From- Modified)
G = nx.Graph()
G.add_edges_from([('A', 'E'), ('B', 'E'), ('E', 'D'), ('C', 'D')])
pos = {'A': [0, 0], 'B': [2, 0], 'C': [4, 2], 'D': [3, 4], 'E': [1, 2]}  # Custom positions

nx.draw(G, pos, with_labels=True, node_size=1500, node_color="skyblue", font_size=15, ax=axes[0, 1])

# Draw branch lengths
for u, v, data in G.edges(data=True):
    # Check if branch_length exists for the current node 'u'
    branch_length = tree[u].get('branch_length') if u in tree and 'branch_length' in tree[u] else None

    if branch_length is not None:
        x1, y1 = pos[u]
        x2, y2 = pos[v]
        axes[0, 1].text((x1 + x2) / 2, (y1 + y2) / 2, f"{branch_length:.1f}",
                        fontsize=10, bbox=dict(facecolor='white', alpha=0.7))

axes[0, 1].set_title('Phylogenetic Tree', fontsize=14)
axes[0, 1].axis('off')

# C. Trait Evolution Visualization (From - Traitgram Style)
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesA'], label='Species A', color='blue')
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesB'], label='Species B', color='orange')
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesC'], label='Species C', color='red')
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesD'], label='Species D', color='purple')
axes[1, 0].set_xlabel('Time', fontsize=12)
axes[1, 0].set_ylabel('Trait Value', fontsize=12)
axes[1, 0].set_title('Trait Evolution Over Time', fontsize=14)
axes[1, 0].legend()

# D. Trait Variation Boxplots (Inspired by, but using final trait values)
final_traits = trait_data[['SpeciesA', 'SpeciesB', 'SpeciesC', 'SpeciesD']].iloc[-1] # Select the last row
sns.boxplot(data=final_traits, ax=axes[1, 1])
axes[1, 1].set_title('Trait Variation at Present', fontsize=14)
axes[1, 1].set_xlabel('Species', fontsize=12)
axes[1, 1].set_ylabel('Final Trait Value', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import networkx as nx
from scipy.cluster.hierarchy import dendrogram, linkage

# Set a random seed for reproducibility
np.random.seed(42)

# --- 1. Time Series Data Simulation ---
timesteps = 500
t = np.arange(timesteps)

# Simulate xt dynamics (Ornstein-Uhlenbeck-like process, but without mentioning it)
def simulate_ou_like(timesteps, mu, theta, sigma, initial_value):
    dt = 1
    values = np.zeros(timesteps)
    values[0] = initial_value
    for i in range(1, timesteps):
        dW = np.random.normal(0, np.sqrt(dt))
        values[i] = values[i-1] + theta * (mu - values[i-1]) * dt + sigma * dW
    return values

xt1 = simulate_ou_like(timesteps, mu=0, theta=0.05, sigma=0.1, initial_value=0)
xt2 = simulate_ou_like(timesteps, mu=0.4, theta=0.1, sigma=0.05, initial_value=0.6)
xt3 = simulate_ou_like(timesteps, mu=-0.2, theta=0.02, sigma=0.08, initial_value=-0.5)

# Simulate theta dynamics (Quadratic relationship with noise)
theta1 = 0.1 + 0.15 * xt1 - 0.05 * xt1**2 - 0.01 * xt1**3 + np.random.normal(scale=0.05, size=timesteps)
theta2 = 0.1 + 0.15 * xt2 - 0.05 * xt2**2 - 0.01 * xt2**3 + np.random.normal(scale=0.05, size=timesteps)
theta3 = 0.1 + 0.15 * xt3 - 0.05 * xt3**2 - 0.01 * xt3**3 + np.random.normal(scale=0.05, size=timesteps)

# Simulate yt dynamics (Sinusoidal with noise)
yt1 = 0.2 * np.sin(0.05*t + xt1) + np.random.normal(scale=0.05, size=timesteps)
yt2 = 0.2 * np.sin(0.05*t + xt2) + np.random.normal(scale=0.05, size=timesteps)
yt3 = 0.2 * np.sin(0.05*t + xt3) + np.random.normal(scale=0.05, size=timesteps)

# Create a list to hold each simulation DataFrame
data_list = []
for var, sim1, sim2, sim3 in zip(['xt', 'theta', 'yt'],
                                  [xt1, theta1, yt1],
                                  [xt2, theta2, yt2],
                                  [xt3, theta3, yt3]):
    for sim_name, sim_data in zip(['Sim1', 'Sim2', 'Sim3'], [sim1, sim2, sim3]):
        df = pd.DataFrame({'Time': t, 'Value': sim_data, 'Variable': var, 'Simulation': sim_name})
        data_list.append(df)
data = pd.concat(data_list, ignore_index=True)

# --- 2. Phylogenetic Tree and Trait Evolution (Image 2) ---
# Simplified Tree Structure
tree = {
    'A': {'to': 'E', 'branch_length': 2.5},
    'B': {'to': 'E', 'branch_length': 2.5},
    'E': {'to': 'D', 'branch_length': 2.3},
    'C': {'to': 'D', 'branch_length': 4.8},
    'D': {}  # Root node
}

# Simulate Trait Evolution (Brownian Motion-like, but without mentioning it)
def simulate_brownian_like(timesteps, sigma, initial_value):
    dt = 1
    values = np.zeros(timesteps)
    values[0] = initial_value
    for i in range(1, timesteps):
        dW = np.random.normal(0, np.sqrt(dt))
        values[i] = values[i-1] + sigma * dW
    return values

trait_timesteps = 500
trait_noise = 0.01

trait_A = simulate_brownian_like(trait_timesteps, sigma=trait_noise, initial_value=0.05)
trait_B = simulate_brownian_like(trait_timesteps, sigma=trait_noise*1.2, initial_value=0.1)  # different sigma
trait_C = simulate_brownian_like(trait_timesteps, sigma=trait_noise*0.8, initial_value=0.2)  # different sigma
trait_D = simulate_brownian_like(trait_timesteps, sigma=trait_noise*1.5, initial_value=0.02)  # different sigma

# Combine trait data into a DataFrame
trait_data = pd.DataFrame({
    'Time': t,
    'SpeciesA': trait_A,
    'SpeciesB': trait_B,
    'SpeciesC': trait_C,
    'SpeciesD': trait_D
})

# --- 3. Species data ---
# Create data set of species features
species_data = pd.DataFrame({
    'Species': ['Sepia elegans', 'Sepia officinalis', 'Illex coindetii', 'Squilla mantis', 'Nephrops norvegicus',
                'Muraena helena', 'Micromesistius poutassou', 'Syngnathus abaster', 'Upeneus pori', 'Pomatomus saltatrix',
                'Knipowitschia caucasica', 'Coryphaena hippurus', 'Trachinotus ovatus', 'Lichia amia', 'Seriola dumerili',
                'Seriola fasciata', 'Scophthalmus rhombus', 'Serranus hepatus', 'Serranus scriba', 'Labrus merula',
                'Labrus viridis', 'Labrus mixtus', 'Symphodus melanocercus', 'Dicentrarchus labrax', 'Siganus rivulatus',
                'Siganus luridus', 'Spicara flexuosa', 'Dentex macrophthalmus', 'Mustelus mustelus'],
    'Trait': np.random.uniform(0.1, 16.38, 29) # Trait values (replace with actual data if available)
})

# --- 4. Visualization (Combining Elements from All Images) ---
fig, axes = plt.subplots(2, 2, figsize=(20, 15)) # Adjust figure size

# A. Time Series Visualization (From Image 1 - Faceted Line Plot)
sns.relplot(x='Time', y='Value', hue='Simulation', col='Variable', kind='line', data=data, ax=axes[0,0], lw=1.5)
axes[0,0].set_title('Time Series Dynamics', fontsize=14)
axes[0,0].set_xlabel('Time', fontsize=12)
axes[0,0].set_ylabel('Value', fontsize=12)
axes[0,0].legend()

# B. Phylogenetic Tree Visualization (From Image 2 - Modified)
G = nx.Graph()
G.add_edges_from([('A', 'E'), ('B', 'E'), ('E', 'D'), ('C', 'D')])
pos = {'A': [0, 0], 'B': [2, 0], 'C': [4, 2], 'D': [3, 4], 'E': [1, 2]}  # Custom positions

nx.draw(G, pos, with_labels=True, node_size=1500, node_color="skyblue", font_size=15, ax=axes[0, 1])

# Draw branch lengths
for u, v, data in G.edges(data=True):
    # Check if branch_length exists for the current node 'u'
    branch_length = tree[u].get('branch_length') if u in tree and 'branch_length' in tree[u] else None

    if branch_length is not None:
        x1, y1 = pos[u]
        x2, y2 = pos[v]
        axes[0, 1].text((x1 + x2) / 2, (y1 + y2) / 2, f"{branch_length:.1f}",
                        fontsize=10, bbox=dict(facecolor='white', alpha=0.7))

axes[0, 1].set_title('Phylogenetic Tree', fontsize=14)
axes[0, 1].axis('off')

# C. Trait Evolution Visualization (From Image 2 - Traitgram Style)
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesA'], label='Species A', color='blue')
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesB'], label='Species B', color='orange')
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesC'], label='Species C', color='red')
axes[1, 0].plot(trait_data['Time'], trait_data['SpeciesD'], label='Species D', color='purple')
axes[1, 0].set_xlabel('Time', fontsize=12)
axes[1, 0].set_ylabel('Trait Value', fontsize=12)
axes[1, 0].set_title('Trait Evolution Over Time', fontsize=14)
axes[1, 0].legend()

# D. Dendrogram based on species trait
Z = linkage(species_data['Trait'].values.reshape(-1, 1), method='ward')
dendrogram(Z, labels=species_data['Species'].tolist(), ax=axes[1, 1], orientation='left')
axes[1, 1].set_title('Hierarchical Clustering of Species by Trait', fontsize=14)
axes[1, 1].set_xlabel('Distance', fontsize=12)
axes[1, 1].set_ylabel('Species', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
# Google Colab Notebook: Adaptive Trait Evolution Modeling

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.integrate import odeint
from scipy.stats import norm, uniform
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C

# Seed for reproducibility
np.random.seed(42)

# --- Simulated Data Generation ---

# Nekton data
def generate_nekton_data():
    log_fecundity = np.random.uniform(4, 16, 100)
    log_longevity = -0.02 * (log_fecundity - 10)**2 + 3 + np.random.normal(0, 0.5, 100)
    return pd.DataFrame({'log_fecundity': log_fecundity, 'log_longevity': log_longevity})

# Carnivore data
def generate_carnivore_data():
    log_body_mass = np.random.uniform(4, 12, 100)
    niche_breadth = -0.05 * (log_body_mass - 8)**2 + 1.5 + np.random.normal(0, 0.2, 100)
    return pd.DataFrame({'log_body_mass': log_body_mass, 'niche_breadth': niche_breadth})

nekton_data = generate_nekton_data()
carnivore_data = generate_carnivore_data()

# --- Polynomial regression fitting (3rd order) ---
from numpy.polynomial.polynomial import Polynomial

x_carn = carnivore_data['log_body_mass']
y_carn = carnivore_data['niche_breadth']
poly_carn = Polynomial.fit(x_carn, y_carn, deg=3)

x_nekton = nekton_data['log_fecundity']
y_nekton = nekton_data['log_longevity']
poly_nekton = Polynomial.fit(x_nekton, y_nekton, deg=3)

# --- Plot polynomial fits ---
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
x_vals = np.linspace(4, 12, 100)
plt.scatter(x_carn, y_carn, alpha=0.3)
plt.plot(x_vals, poly_carn(x_vals), color='r')
plt.title('Carnivore Data Polynomial Fit')

plt.subplot(1, 2, 2)
x_vals2 = np.linspace(4, 16, 100)
plt.scatter(x_nekton, y_nekton, alpha=0.3)
plt.plot(x_vals2, poly_nekton(x_vals2), color='g')
plt.title('Nekton Data Polynomial Fit')
plt.show()

# --- Stochastic Differential Equation simulation (OU process) ---
def simulate_ou_process(mu, theta, sigma, initial, timesteps=500):
    dt = 1
    x = np.zeros(timesteps)
    x[0] = initial
    for i in range(1, timesteps):
        dW = np.random.normal(0, np.sqrt(dt))
        x[i] = x[i-1] + theta * (mu - x[i-1]) * dt + sigma * dW
    return x

xt = simulate_ou_process(mu=0, theta=0.1, sigma=0.05, initial=0)

# --- Reaction-diffusion simulation ---
def reaction_diffusion(y0, t, D, reaction_func):
    dydt = D * np.gradient(np.gradient(y0)) + reaction_func(y0)
    return dydt

# Example reaction function
reaction_func = lambda y: 0.1 * y * (1 - y)

# --- Gaussian Process Regression ---
kernel = C(1.0, (1e-3, 1e3)) * RBF(10, (1e-2, 1e2))
gp = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=9)

X_gp = x_carn.values.reshape(-1, 1)
Y_gp = y_carn.values.reshape(-1, 1)
gp.fit(X_gp, Y_gp)

X_test = np.linspace(4, 12, 100).reshape(-1, 1)
Y_pred, sigma = gp.predict(X_test, return_std=True)

plt.figure(figsize=(8, 5))
plt.scatter(X_gp, Y_gp, alpha=0.3)
plt.plot(X_test, Y_pred, color='purple')
plt.fill_between(X_test.flatten(), Y_pred.flatten()-1.96*sigma, Y_pred.flatten()+1.96*sigma, color='purple', alpha=0.2)
plt.title('Gaussian Process Regression - Carnivore Niche Breadth')
plt.show()

# --- Approximate Bayesian Computation (ABC) ---
def simulate_summary_statistics(params):
    alpha, beta = params
    simulated_values = alpha * np.random.normal(0, 1, 100) + beta
    return simulated_values.mean(), simulated_values.std()

def distance(sim_stats, obs_stats):
    return np.sqrt(np.sum((np.array(sim_stats) - np.array(obs_stats)) ** 2))

obs_data = y_carn.values
obs_stats = (obs_data.mean(), obs_data.std())

accepted_params = []
for _ in range(5000):
    candidate_alpha = uniform.rvs(0, 2)
    candidate_beta = norm.rvs(0, 1)
    sim_stats = simulate_summary_statistics((candidate_alpha, candidate_beta))
    dist = distance(sim_stats, obs_stats)
    if dist < 0.1:
        accepted_params.append((candidate_alpha, candidate_beta))

accepted_params = np.array(accepted_params)

plt.figure(figsize=(10, 5))
plt.scatter(accepted_params[:,0], accepted_params[:,1], alpha=0.3)
plt.xlabel('Alpha')
plt.ylabel('Beta')
plt.title('Posterior Parameter Distribution (ABC)')
plt.show()

# --- Conclusion: Results visualization and discussion placeholder ---
print("The polynomial regression, Gaussian process modeling, and ABC-based inference have been completed following the paper's methodology.")
```

```python
# Google Colab Code: Adaptive Trait Evolution Modeling

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from scipy.stats import uniform, norm
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C

# Sample trait data (replace with your actual data input if needed)
t = np.linspace(0, 10, 100)
trait_A = np.sin(t) + 0.1*np.random.randn(100)
trait_B = np.cos(t) + 0.1*np.random.randn(100)
trait_C = np.sin(t/2) + 0.1*np.random.randn(100)
trait_D = np.cos(t/2) + 0.1*np.random.randn(100)

trait_data = pd.DataFrame({
    'Time': t,
    'SpeciesA': trait_A,
    'SpeciesB': trait_B,
    'SpeciesC': trait_C,
    'SpeciesD': trait_D
})

# Ornstein–Uhlenbeck SDE simulation
def simulate_OU_process(alpha, theta, sigma, y0, t):
    y = np.zeros(len(t))
    y[0] = y0
    dt = t[1] - t[0]
    for i in range(1, len(t)):
        dy = alpha * (theta - y[i-1]) * dt + sigma * np.random.normal(0, np.sqrt(dt))
        y[i] = y[i-1] + dy
    return y

# Polynomial optimal adaptive function (order 4)
def theta_xt(beta, xt):
    return sum([beta[i]*xt**i for i in range(5)])

# Simulate trait evolution with OU SDE driven by polynomial theta_xt
def simulate_trait(alpha_y, sigma_y, beta, xt, t, y0):
    y = np.zeros(len(t))
    y[0] = y0
    dt = t[1] - t[0]
    for i in range(1, len(t)):
        theta_t = theta_xt(beta, xt[i])
        dy = alpha_y * (theta_t - y[i-1]) * dt + sigma_y * np.random.normal(0, np.sqrt(dt))
        y[i] = y[i-1] + dy
    return y

# Approximate Bayesian Computation (ABC) function with wider priors and relaxed tolerance
def abc_simulation(trait_obs, xt, t, prior_samples=3000, tolerance=10.0):
    accepted_params = []
    for _ in range(prior_samples):
        alpha_y = uniform.rvs(0.05, 1.5)  # Wider range
        sigma_y = uniform.rvs(0.05, 2.0)  # Larger noise allowance
        beta = norm.rvs(loc=[0, 0.5, 0.1, -0.05, 0.6], scale=[1.0, 0.5, 0.1, 0.1, 0.3])

        y_sim = simulate_trait(alpha_y, sigma_y, beta, xt, t, trait_obs[0])
        dist = np.linalg.norm((y_sim - trait_obs) / (np.std(trait_obs) + 1e-5))  # relative distance

        if dist < tolerance:
            accepted_params.append((alpha_y, sigma_y, beta))

    return accepted_params

# Gaussian Process Regression surrogate model fitting
def gpr_surrogate(X_train, y_train):
    kernel = C(1.0) * RBF(length_scale=1.0)
    gp = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=5, alpha=0.01)
    gp.fit(X_train, y_train)
    return gp

# Example: Simulate and infer for Species A
xt = simulate_OU_process(alpha=0.3, theta=0, sigma=1.5, y0=0, t=t)
accepted_parameters = abc_simulation(trait_data['SpeciesA'].values, xt, t, tolerance=10.0)

# Visualize results
print(f"Accepted {len(accepted_parameters)} parameter sets")

# Plot simulated best-fit from median parameters if available
if accepted_parameters:
    median_idx = len(accepted_parameters) // 2
    alpha_y, sigma_y, beta = accepted_parameters[median_idx]
    simulated_y = simulate_trait(alpha_y, sigma_y, beta, xt, t, trait_data['SpeciesA'].values[0])

    plt.figure(figsize=(10, 6))
    plt.plot(t, trait_data['SpeciesA'], label='Observed Species A')
    plt.plot(t, simulated_y, label='Simulated (Median Fit)', linestyle='dashed')
    plt.xlabel('Time')
    plt.ylabel('Trait Value')
    plt.legend()
    plt.title('Adaptive Trait Evolution Simulation vs Observed (Species A)')
    plt.show()

# Export accepted parameter sets for GPR surrogate modeling
if accepted_parameters:
    param_arr = np.array([[p[0], p[1]] + list(p[2]) for p in accepted_parameters])
    summary_stat = param_arr[:,0] + param_arr[:,1]  # example summary: alpha + sigma

    gp = gpr_surrogate(param_arr, summary_stat)

    # Predict with GPR on new test points
    test_samples = param_arr[np.random.choice(range(param_arr.shape[0]), size=20, replace=False)]
    predicted_stats, std_dev = gp.predict(test_samples, return_std=True)

    plt.errorbar(range(len(predicted_stats)), predicted_stats, yerr=std_dev, fmt='o')
    plt.title('GPR Surrogate Model Predictions on Parameter Space')
    plt.ylabel('Predicted Summary Statistic')
    plt.xlabel('Test Sample Index')
    plt.show()
```

```python
# -*- coding: utf-8 -*-
"""Trait_Evolution_Simulation_Expanded.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/1eri-WzD-cpBhqCg1oomQTOXT4wJkXJ_B

## Trait Evolution Simulation in Python (Expanded)

This Colab notebook implements various models for simulating trait evolution,
based on mathematical formulations described in the provided paper. It
includes:

*   **Ordinary Differential Equation (ODE) model:** Deterministic trait
    evolution.
*   **Stochastic Differential Equation (SDE) model:** Incorporates
    randomness.
*   **Covariate Modeling:** Brownian motion and Ornstein-Uhlenbeck
    processes.
*   **Phylogenetic Tree Simulation:** Simulates trait evolution along a tree.
*   **Fourier Series Regression**
*   **Wavelet Transforms**
*   **Neural Network Regression**
*   **Multi-Dimensional Evolutionary Models (coupled ODEs)**
*   **Spatial Evolution and Reaction-Diffusion Models**
*   **Hamiltonian Mechanics for Evolutionary Constraints**
*   **Optimal Trait and Covariate Relationship (polynomial expansion)**
*   **Covariate Modeling (Brownian Motion and Ornstein-Uhlenbeck)**
*   **Lagrange Polynomial Interpolation**

This code avoids the use of CSV files, generating data directly within the
script.

"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.fft import fft, ifft
import pywt  # For wavelet transforms
import tensorflow as tf  # For neural networks
from sklearn.preprocessing import MinMaxScaler #For Neural Network
from sklearn.model_selection import train_test_split
from sympy import * # For Hamiltonian Mechanics
from scipy.interpolate import lagrange

## 1. Ordinary Differential Equation (ODE) Model

def trait_evolution_ode(t, y, x, t_val):
  """
  Defines the ODE for trait evolution.

  Args:
    t: Time.
    y: Trait value.
    x: Environmental covariate (constant for simplicity in this example).
    t_val: Time value

  Returns:
    The derivative of the trait value with respect to time.
  """
  f = 0.1 * (1 - y)  # Example: Trait evolves towards an optimal value of 1
  return f

def simulate_ode(x, t_end=10, y0=0.0):
  """
  Simulates the trait evolution using the ODE.

  Args:
    x: Environmental covariate value.
    t_end: End time for the simulation.
    y0: Initial trait value.

  Returns:
    A tuple containing the time points and the trait values.
  """
  t_span = [0, t_end]
  t_eval = np.linspace(0, t_end, 100) # Evaluate the solution at these points
  sol = solve_ivp(trait_evolution_ode, t_span, [y0], args=(x, t_eval), dense_output=True, t_eval=t_eval)
  return sol.t, sol.y[0]

# Example Usage of ODE
x_val = 0.5 #Environmental covariate value
time_ode, trait_ode = simulate_ode(x_val)

# Plotting the ODE results
plt.figure(figsize=(8, 6))
plt.plot(time_ode, trait_ode, label='ODE Trait Evolution')
plt.xlabel('Time')
plt.ylabel('Trait Value')
plt.title('Deterministic Trait Evolution (ODE)')
plt.legend()
plt.grid(True)
plt.show()

## 2. Stochastic Differential Equation (SDE) Model

def trait_evolution_sde(y0, alpha, theta, sigma, dt, num_steps):
    """
    Simulates trait evolution using the SDE.

    Args:
        y0: Initial trait value.
        alpha: Selective force coefficient.
        theta: Optimal trait value.
        sigma: Volatility.
        dt: Time step.
        num_steps: Number of simulation steps.

    Returns:
        A tuple containing the time points and the trait values.
    """
    y = np.zeros(num_steps + 1)
    y[0] = y0
    t = np.linspace(0, dt * num_steps, num_steps + 1)

    for i in range(num_steps):
        dW = np.random.normal(0, np.sqrt(dt))  # Wiener process increment
        y[i+1] = y[i] + alpha * (theta - y[i]) * dt + sigma * dW

    return t, y

# Example Usage of SDE
y0_sde = 0.0  # Initial trait value
alpha_sde = 0.1  # Selective force coefficient
theta_sde = 1.0  # Optimal trait value
sigma_sde = 0.1  # Volatility
dt_sde = 0.1  # Time step
num_steps_sde = 100

time_sde, trait_sde = trait_evolution_sde(y0_sde, alpha_sde, theta_sde, sigma_sde, dt_sde, num_steps_sde)

# Plotting the SDE results
plt.figure(figsize=(8, 6))
plt.plot(time_sde, trait_sde, label='SDE Trait Evolution')
plt.xlabel('Time')
plt.ylabel('Trait Value')
plt.title('Stochastic Trait Evolution (SDE)')
plt.legend()
plt.grid(True)
plt.show()

## 3. Covariate Modeling

def brownian_motion(sigma_x, dt, num_steps):
    """
    Simulates a Brownian motion covariate.

    Args:
        sigma_x: Volatility of the Brownian motion.
        dt: Time step.
        num_steps: Number of simulation steps.

    Returns:
        A tuple containing the time points and the covariate values.
    """
    x = np.zeros(num_steps + 1)
    t = np.linspace(0, dt * num_steps, num_steps + 1)

    for i in range(num_steps):
        dW = np.random.normal(0, np.sqrt(dt))
        x[i+1] = x[i] + sigma_x * dW

    return t, x

def ornstein_uhlenbeck(x0, alpha_x, theta_x, sigma_x, dt, num_steps):
    """
    Simulates an Ornstein-Uhlenbeck process covariate.

    Args:
        x0: Initial covariate value.
        alpha_x: Mean reversion strength.
        theta_x: Long-term mean.
        sigma_x: Volatility.
        dt: Time step.
        num_steps: Number of simulation steps.

    Returns:
        A tuple containing the time points and the covariate values.
    """
    x = np.zeros(num_steps + 1)
    x[0] = x0
    t = np.linspace(0, dt * num_steps, num_steps + 1)

    for i in range(num_steps):
        dW = np.random.normal(0, np.sqrt(dt))
        x[i+1] = x[i] + alpha_x * (theta_x - x[i]) * dt + sigma_x * dW

    return t, x

# Example Usage of Covariate Models
dt_cov = 0.1
num_steps_cov = 100

# Brownian Motion
sigma_x = 0.1
time_bm, covariate_bm = brownian_motion(sigma_x, dt_cov, num_steps_cov)

# Ornstein-Uhlenbeck
x0_ou = 0.0
alpha_x = 0.2
theta_x = 1.0
sigma_x = 0.1
time_ou, covariate_ou = ornstein_uhlenbeck(x0_ou, alpha_x, theta_x, sigma_x, dt_cov, num_steps_cov)

# Plotting the Covariate Models
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(time_bm, covariate_bm, label='Brownian Motion Covariate')
plt.xlabel('Time')
plt.ylabel('Covariate Value')
plt.title('Brownian Motion Covariate')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(time_ou, covariate_ou, label='Ornstein-Uhlenbeck Covariate')
plt.xlabel('Time')
plt.ylabel('Covariate Value')
plt.title('Ornstein-Uhlenbeck Covariate')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

## 4. Phylogenetic Tree Simulation (Simplified)

def simulate_phylogenetic_tree(num_species, alpha, beta, sigma, t_total):
    """
    Simulates trait evolution in a simplified phylogenetic tree.

    Args:
        num_species: Number of species in the tree.
        alpha: Selective force coefficient.
        beta: Regression parameter for optimal trait-covariate relationship.
        sigma: Volatility.
        t_total: Total evolutionary time.

    Returns:
        A dictionary containing the trait values for each species.
    """

    trait_values = {}
    np.random.seed(42)  # for reproducibility

    for i in range(num_species):
        # Generate a random evolutionary time for each species
        t_a = np.random.uniform(0, t_total)

        # Simulate the covariate at time t_a (using a simplified approach)
        x_a = np.random.normal(0, 1)  # Simplified covariate value

        # Calculate the optimal trait value
        theta_a = beta * x_a

        # Simulate the trait value
        y_ik = np.exp(-alpha * t_a) + alpha * np.exp(-alpha * t_a) * beta * integrate_exponential(alpha, t_a, x_a) + sigma * np.random.normal(0, np.sqrt(t_a))
        trait_values[f"species_{i}"] = y_ik

    return trait_values

def integrate_exponential(alpha, t_a, x_a):
    """
    Calculates the integral of exp(alpha * s) * x_a ds from 0 to t_a.
    """
    return x_a * (np.exp(alpha * t_a) - 1) / alpha

# Example Usage of Phylogenetic Tree Simulation
num_species = 5
alpha_phylo = 0.1
beta_phylo = 0.5
sigma_phylo = 0.1
t_total = 10

trait_values = simulate_phylogenetic_tree(num_species, alpha_phylo, beta_phylo, sigma_phylo, t_total)

# Print the results
print("Trait values for each species:")
for species, trait_value in trait_values.items():
    print(f"{species}: {trait_value:.4f}")

# Visualization (Bar plot of trait values)
species_names = list(trait_values.keys())
trait_values_list = list(trait_values.values())

plt.figure(figsize=(10, 6))
plt.bar(species_names, trait_values_list, color='skyblue')
plt.xlabel('Species')
plt.ylabel('Trait Value')
plt.title('Simulated Trait Values in Phylogenetic Tree')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for readability
plt.tight_layout()
plt.show()

## 5. Fourier Series Regression

def fourier_series(x, a0, a, b, n):
    """
    Calculates the Fourier series.

    Args:
        x: Input values.
        a0: Constant term.
        a: Cosine coefficients.
        b: Sine coefficients.
        n: Number of terms.

    Returns:
        The Fourier series values.
    """
    result = a0
    for i in range(1, n + 1):
        result += a[i-1] * np.cos(i * x) + b[i-1] * np.sin(i * x)
    return result

# Example Usage of Fourier Series Regression
# Generate some periodic data (e.g., a sine wave)
x_fourier = np.linspace(-np.pi, np.pi, 200)
y_fourier = np.sin(2 * x_fourier) + 0.5 * np.cos(3 * x_fourier) + 0.1 * np.random.randn(200)

# Calculate Fourier coefficients (simplified)
n_terms = 5
a0 = np.mean(y_fourier)
a = np.zeros(n_terms)
b = np.zeros(n_terms)
for i in range(1, n_terms + 1):
    a[i-1] = np.mean(y_fourier * np.cos(i * x_fourier))
    b[i-1] = np.mean(y_fourier * np.sin(i * x_fourier))

# Calculate the Fourier series approximation
y_approx_fourier = fourier_series(x_fourier, a0, a, b, n_terms)

# Plotting the results
plt.figure(figsize=(8, 6))
plt.plot(x_fourier, y_fourier, label='Original Data')
plt.plot(x_fourier, y_approx_fourier, label='Fourier Series Approximation')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Fourier Series Regression')
plt.legend()
plt.grid(True)
plt.show()

## 6. Wavelet Transforms

def wavelet_transform(data, wavelet='haar', level=3):
    """
    Applies a discrete wavelet transform.

    Args:
        data: Input data.
        wavelet: Wavelet type (e.g., 'haar', 'db1').
        level: Decomposition level.

    Returns:
        The wavelet coefficients.
    """
    coeffs = pywt.wavedec(data, wavelet, level=level)
    return coeffs

# Example Usage of Wavelet Transforms
# Generate some data (e.g., a step function)
x_wavelet = np.linspace(0, 1, 200)
y_wavelet = np.zeros(200)
y_wavelet[50:150] = 1
y_wavelet += 0.1 * np.random.randn(200)

# Apply wavelet transform
wavelet = 'db1'  # Daubechies wavelet of order 1
coeffs = wavelet_transform(y_wavelet, wavelet=wavelet, level=3)

# Plotting the results (original data and approximation)
plt.figure(figsize=(8, 6))
plt.plot(x_wavelet, y_wavelet, label='Original Data')

# Reconstruct the signal from the wavelet coefficients (approximation)
y_reconstructed = pywt.waverec(coeffs, wavelet)
plt.plot(x_wavelet, y_reconstructed[:200], label='Wavelet Approximation')  # Truncate to match original length
plt.xlabel('x')
plt.ylabel('y')
plt.title('Wavelet Transform')
plt.legend()
plt.grid(True)
plt.show()

## 7. Neural Network Regression

def create_nn_model(input_shape):
    """
    Creates a simple neural network model.

    Args:
        input_shape: Shape of the input data.

    Returns:
        A compiled neural network model.
    """
    model = tf.keras.models.Sequential([
        tf.keras.layers.Dense(64, activation='relu', input_shape=input_shape),
        tf.keras.layers.Dense(64, activation='relu'),
        tf.keras.layers.Dense(1)  # Output layer (single value)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model

# Example Usage of Neural Network Regression
# Generate some non-linear data
x_nn = np.linspace(-5, 5, 200)
y_nn = x_nn**3 + 2*x_nn**2 - 5*x_nn + 3 + 10 * np.random.randn(200)

# Reshape the data for the neural network
x_nn = x_nn.reshape(-1, 1)  # Reshape to (200, 1)
y_nn = y_nn.reshape(-1, 1)

# Scale the data using MinMaxScaler
x_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()

x_scaled = x_scaler.fit_transform(x_nn)
y_scaled = y_scaler.fit_transform(y_nn)

# Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y_scaled, test_size=0.2, random_state=42)

# Create the neural network model
model = create_nn_model(input_shape=(1,))

# Train the model
model.fit(x_train, y_train, epochs=50, batch_size=32, verbose=0)

# Evaluate the model
loss = model.evaluate(x_test, y_test, verbose=0)
print(f"Neural Network Loss: {loss:.4f}")

# Make predictions
y_pred_scaled = model.predict(x_test)

# Inverse transform the scaled predictions
y_pred = y_scaler.inverse_transform(y_pred_scaled)

# Plotting the results
plt.figure(figsize=(8, 6))
plt.scatter(x_test, y_test, label='Original Data', color='blue')
plt.scatter(x_test, y_pred, label='Neural Network Prediction', color='red', alpha=0.7)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Neural Network Regression')
plt.legend()
plt.grid(True)
plt.show()

## 8. Multi-Dimensional Evolutionary Models (coupled ODEs)

def coupled_odes(t, y, x):
    """
    Defines a system of coupled ODEs for trait evolution.

    Args:
        t: Time.
        y: Trait values (array of length n).
        x: Environmental covariate.

    Returns:
        The derivatives of the trait values with respect to time.
    """
    y1, y2 = y
    f1 = 0.1 * (1 - y1) + 0.05 * y2  # Trait 1 depends on Trait 2
    f2 = 0.2 * (0.5 - y2) - 0.1 * y1  # Trait 2 depends on Trait 1
    return [f1, f2]

def simulate_coupled_odes(x, t_end=10, y0=[0.0, 0.0]):
    """
    Simulates the coupled ODEs.

    Args:
        x: Environmental covariate.
        t_end: End time.
        y0: Initial trait values (list of length n).

    Returns:
        A tuple containing the time points and the trait values.
    """
    t_span = [0, t_end]
    t_eval = np.linspace(0, t_end, 100)
    sol = solve_ivp(coupled_odes, t_span, y0, args=(x,), dense_output=True, t_eval=t_eval)
    return sol.t, sol.y

# Example Usage of Coupled ODEs
x_coupled = 0.5
time_coupled, traits_coupled = simulate_coupled_odes(x_coupled)

# Plotting the results
plt.figure(figsize=(8, 6))
plt.plot(time_coupled, traits_coupled[0], label='Trait 1')
plt.plot(time_coupled, traits_coupled[1], label='Trait 2')
plt.xlabel('Time')
plt.ylabel('Trait Value')
plt.title('Coupled ODEs')
plt.legend()
plt.grid(True)
plt.show()

## 9. Spatial Evolution and Reaction-Diffusion Models (Simplified)

def reaction_diffusion(y, t, D, f):
    """
    Defines the reaction-diffusion equation in 1D.

    Args:
        y: Trait values at each spatial point.
        t: Time.
        D: Diffusion coefficient.
        f: Reaction term (function of y).

    Returns:
        The derivatives of the trait values with respect to time.
    """
    dydt = np.zeros_like(y)
    for i in range(1, len(y) - 1):
        dydt[i] = D * (y[i+1] - 2*y[i] + y[i-1]) + f(y[i])
    # Boundary conditions (zero flux)
    dydt[0] = D * (y[1] - y[0]) + f(y[0])
    dydt[-1] = D * (y[-2] - y[-1]) + f(y[-1])
    return dydt

def simulate_reaction_diffusion(D, f, y0, t_end=10, num_points=50):
    """
    Simulates the reaction-diffusion equation.

    Args:
        D: Diffusion coefficient.
        f: Reaction term (function).
        y0: Initial trait values (array of length num_points).
        t_end: End time.
        num_points: Number of spatial points.

    Returns:
        A tuple containing the time points and the trait values.
    """
    t_span = [0, t_end]
    t_eval = np.linspace(0, t_end, 50)
    sol = solve_ivp(lambda t, y: reaction_diffusion(y, t, D, f), t_span, y0, dense_output=True, t_eval=t_eval)
    return sol.t, sol.y

# Example Usage of Reaction-Diffusion
D = 0.01  # Diffusion coefficient
f = lambda y: 0.1 * y * (1 - y)  # Logistic growth (reaction term)
num_points = 50
y0 = np.random.rand(num_points)  # Initial random trait values

time_rd, traits_rd = simulate_reaction_diffusion(D, f, y0)

# Plotting the results (trait values at the final time point)
plt.figure(figsize=(8, 6))
plt.plot(np.linspace(0, 1, num_points), traits_rd[:, -1], label='Trait Value at t=t_end')
plt.xlabel('Spatial Position')
plt.ylabel('Trait Value')
plt.title('Reaction-Diffusion Model')
plt.legend()
plt.grid(True)
plt.show()

## 10. Hamiltonian Mechanics for Evolutionary Constraints (Simplified)

# Define symbols
y, y_dot, x = symbols('y y_dot x')

# Define potential function (example: quadratic potential)
V = 0.5 * y**2 + 0.1 * x * y

# Define Hamiltonian
H = 0.5 * y_dot**2 + V

# Calculate equations of motion (Hamilton's equations)
dH_dy_dot = diff(H, y_dot)
dH_dy = diff(H, y)

# Print the equations of motion
print("Hamiltonian:", H)
print("dH/dy_dot:", dH_dy_dot)
print("dH/dy:", dH_dy)

## 11. Optimal Trait and Covariate Relationship (polynomial expansion)
def optimal_trait(x, beta):
    """
    Calculates the optimal trait value based on a polynomial expansion.

    Args:
        x: Covariate value.
        beta: Regression parameters (list or array).

    Returns:
        The optimal trait value.
    """
    theta = 0
    for j, b in enumerate(beta):
        theta += b * (x**j)
    return theta

# Example Usage of Polynomial Expansion
x_poly = np.linspace(-2, 2, 100)
beta = [0.5, 1.0, -0.2]  # Regression parameters
theta_poly = optimal_trait(x_poly, beta)

# Plotting the results
plt.figure(figsize=(8, 6))
plt.plot(x_poly, theta_poly, label='Optimal Trait')
plt.xlabel('Covariate Value (x)')
plt.ylabel('Optimal Trait Value (theta)')
plt.title('Optimal Trait as a Polynomial Function of Covariate')
plt.legend()
plt.grid(True)
plt.show()

## 12. Lagrange Polynomial Interpolation
def lagrange_interpolation(x, x_points, y_points):
    """
    Performs Lagrange interpolation.

    Args:
        x: The point at which to interpolate.
        x_points: The x-coordinates of the data points.
        y_points: The y-coordinates of the data points.

    Returns:
        The interpolated value at x.
    """
    n = len(x_points)
    result = 0.0
    for i in range(n):
        term = y_points[i]
        for j in range(n):
            if i != j:
                term *= (x - x_points[j]) / (x_points[i] - x_points[j])
        result += term
    return result

# Example Usage of Lagrange Interpolation
x_points = np.array([0, 1, 2, 3])
y_points = np.array([1, 3, 2, 4])
x_interp = np.linspace(0, 3, 100)
y_interp = [lagrange_interpolation(x, x_points, y_points) for x in x_interp]

# Plotting the results
plt.figure(figsize=(8, 6))
plt.plot(x_interp, y_interp, label='Lagrange Interpolation')
plt.scatter(x_points, y_points, color='red', label='Data Points')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Lagrange Polynomial Interpolation')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp, simpson
from scipy.interpolate import lagrange

# 1. Evolution of interdependent traits (ODE system)
def ode_system(y, t):
    y1, y2 = y
    dy1dt = y1 * (1 - y2)
    dy2dt = y2 * (y1 - 1)
    return [dy1dt, dy2dt]

t = np.linspace(0, 10, 200)
y0 = [1.0, 0.5]
sol = odeint(ode_system, y0, t)

# 2. Spatial evolution (reaction-diffusion)
x = np.linspace(0, 10, 100)
y_spatial = np.exp(-((x - 5) ** 2))
D = 0.1
laplacian = np.gradient(np.gradient(y_spatial, x), x)
reaction = y_spatial * (1 - y_spatial)
dy_dt = D * laplacian + reaction

# 3. Hamiltonian evolution
m = 1.0
y = np.linspace(-2, 2, 100)
V = y**2
H = 0.5 * m * (y**2) + V

# 4. Stochastic differential equation
np.random.seed(42)
dt = 0.01
T = 10
N = int(T/dt)
t_sde = np.linspace(0, T, N)
alpha = 0.5
theta = 1.0
sigma = 0.3
Wt = np.cumsum(np.sqrt(dt) * np.random.randn(N))
yt = np.zeros(N)
for i in range(1, N):
    dyt = alpha * (theta - yt[i-1]) * dt + sigma * (Wt[i] - Wt[i-1])
    yt[i] = yt[i-1] + dyt

# 5. Optimal trait as polynomial function of covariate
x_poly = np.linspace(0, 5, 100)
betas = [1, -0.5, 0.1]
theta_poly = sum(b * x_poly**i for i, b in enumerate(betas))

# 6. Covariate with Brownian and OU processes
W_x = np.cumsum(np.sqrt(dt) * np.random.randn(N))
x_bm = sigma * W_x
x_ou = np.zeros(N)
x0 = 0.0
for i in range(1, N):
    dx = alpha * (theta - x_ou[i-1]) * dt + sigma * (W_x[i] - W_x[i-1])
    x_ou[i] = x_ou[i-1] + dx

# 7. Expected integrals
alpha_y = 0.1
E_I2 = 2 * simps(np.exp(2*alpha_y*t) * t, t)
E_I4 = 4 * simps(np.exp(2*alpha_y*t) * t**2, t)

# 8. Lagrange interpolation
xi = [0, 1, 2]
yi = [1, 3, 2]
poly = lagrange(xi, yi)
x_interp = np.linspace(0, 2, 100)
y_interp = poly(x_interp)

# 9. Phylogenetic tree trait simulation
ta = 5
tree_t = np.linspace(0, ta, 200)
phi_integral = simps(np.exp(0.5 * tree_t) * tree_t, tree_t)
yik = np.exp(-alpha * ta) * 0.5 + alpha * np.exp(-alpha * ta) * phi_integral + sigma * np.sum(np.exp(-alpha * (ta - tree_t)) * np.random.randn(200)) * np.sqrt(dt)

# Visualization
fig, axs = plt.subplots(3, 3, figsize=(15, 12))
axs[0,0].plot(t, sol[:,0], label='y1')
axs[0,0].plot(t, sol[:,1], label='y2')
axs[0,0].set_title('Interdependent Traits (ODE)')
axs[0,0].legend()

axs[0,1].plot(x, dy_dt)
axs[0,1].set_title('Spatial Reaction-Diffusion')

axs[0,2].plot(y, H)
axs[0,2].set_title('Hamiltonian Evolution')

axs[1,0].plot(t_sde, yt)
axs[1,0].set_title('Stochastic Trait Evolution (SDE)')

axs[1,1].plot(x_poly, theta_poly)
axs[1,1].set_title('Polynomial Optimal Trait')

axs[1,2].plot(t_sde, x_bm, label='Brownian')
axs[1,2].plot(t_sde, x_ou, label='Ornstein-Uhlenbeck')
axs[1,2].set_title('Covariate Models')
axs[1,2].legend()

axs[2,0].bar(['E[I2]', 'E[I4]'], [E_I2, E_I4])
axs[2,0].set_title('Expected Evolutionary Integrals')

axs[2,1].plot(x_interp, y_interp)
axs[2,1].scatter(xi, yi, color='red')
axs[2,1].set_title('Lagrange Interpolation')

axs[2,2].bar(['yik'], [yik])
axs[2,2].set_title('Trait Evolution on Phylogenetic Node')

plt.tight_layout()
plt.show()
```

```python
"""
Adaptive Trait Evolution Modeling
Scientific Visualization Version
---------------------------------
This script reproduces:
1. Trait–fitness landscape estimation
2. Polynomial and Gaussian Process regression
3. Ornstein–Uhlenbeck evolutionary dynamics
4. Approximate Bayesian Computation inference

All figures are formatted for scientific publication.
"""

# ============================================================
# 1. IMPORTS AND GLOBAL STYLE CONFIGURATION
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm, uniform
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel

# Reproducibility
np.random.seed(42)

# Matplotlib publication settings
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12,
    "axes.labelsize": 12,
    "axes.titlesize": 13,
    "figure.dpi": 300,
    "savefig.dpi": 600,
    "axes.spines.top": False,
    "axes.spines.right": False
})

# ============================================================
# 2. SIMULATED ECOLOGICAL DATA
# ============================================================

def generate_nekton():
    x = np.random.uniform(4, 16, 150)
    y = -0.02 * (x - 10)**2 + 3 + np.random.normal(0, 0.5, len(x))
    return pd.DataFrame({"log_fecundity": x, "log_longevity": y})

def generate_carnivore():
    x = np.random.uniform(4, 12, 150)
    y = -0.05 * (x - 8)**2 + 1.5 + np.random.normal(0, 0.2, len(x))
    return pd.DataFrame({"log_body_mass": x, "niche_breadth": y})

nekton = generate_nekton()
carnivore = generate_carnivore()

# ============================================================
# 3. TRAIT–FITNESS LANDSCAPE ESTIMATION (POLYNOMIAL MODEL)
# ============================================================

from numpy.polynomial.polynomial import Polynomial

poly_c = Polynomial.fit(carnivore.log_body_mass,
                        carnivore.niche_breadth, deg=3)

poly_n = Polynomial.fit(nekton.log_fecundity,
                        nekton.log_longevity, deg=3)

# ============================================================
# 4. GAUSSIAN PROCESS REGRESSION
# ============================================================

kernel = ConstantKernel(1.0) * RBF(length_scale=5.0)
gp = GaussianProcessRegressor(kernel=kernel,
                              n_restarts_optimizer=5)

X_train = carnivore.log_body_mass.values.reshape(-1, 1)
y_train = carnivore.niche_breadth.values
gp.fit(X_train, y_train)

X_pred = np.linspace(4, 12, 200).reshape(-1, 1)
y_pred, y_std = gp.predict(X_pred, return_std=True)

# ============================================================
# 5. ORNSTEIN–UHLENBECK EVOLUTIONARY DYNAMICS
# ============================================================

def simulate_ou(mu, theta, sigma, x0, steps=500):
    dt = 1.0
    x = np.zeros(steps)
    x[0] = x0
    for t in range(1, steps):
        dW = np.random.normal(0, np.sqrt(dt))
        x[t] = x[t-1] + theta*(mu - x[t-1])*dt + sigma*dW
    return x

ou_path = simulate_ou(mu=0.0, theta=0.1, sigma=0.05, x0=0)

# ============================================================
# 6. APPROXIMATE BAYESIAN COMPUTATION (ABC)
# ============================================================

def simulate_stats(alpha, beta):
    sim = alpha * np.random.normal(0, 1, 200) + beta
    return sim.mean(), sim.std()

def euclidean_distance(a, b):
    return np.sqrt(np.sum((np.array(a) - np.array(b))**2))

observed = (y_train.mean(), y_train.std())
accepted = []

for _ in range(6000):
    a = uniform.rvs(0, 2)
    b = norm.rvs(0, 1)
    sim = simulate_stats(a, b)
    if euclidean_distance(sim, observed) < 0.1:
        accepted.append((a, b))

accepted = np.array(accepted)

# ============================================================
# 7. PUBLICATION-QUALITY MULTIPANEL FIGURE
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# (A) Carnivore Polynomial Fit
x_plot = np.linspace(4, 12, 200)
axes[0,0].scatter(carnivore.log_body_mass,
                  carnivore.niche_breadth,
                  s=15, alpha=0.4)
axes[0,0].plot(x_plot, poly_c(x_plot), linewidth=2)
axes[0,0].set_title("A. Polynomial Fit: Carnivore Trait Landscape")
axes[0,0].set_xlabel("Log Body Mass")
axes[0,0].set_ylabel("Niche Breadth")

# (B) Gaussian Process with 95% CI
axes[0,1].scatter(X_train, y_train, s=15, alpha=0.3)
axes[0,1].plot(X_pred, y_pred, linewidth=2)
axes[0,1].fill_between(
    X_pred.flatten(),
    y_pred - 1.96*y_std,
    y_pred + 1.96*y_std,
    alpha=0.2
)
axes[0,1].set_title("B. Gaussian Process Regression")
axes[0,1].set_xlabel("Log Body Mass")
axes[0,1].set_ylabel("Predicted Niche Breadth")

# (C) OU Evolutionary Trajectory
axes[1,0].plot(ou_path, linewidth=1.5)
axes[1,0].set_title("C. Ornstein–Uhlenbeck Trait Evolution")
axes[1,0].set_xlabel("Time")
axes[1,0].set_ylabel("Trait Value")

# (D) ABC Posterior Distribution
axes[1,1].scatter(accepted[:,0], accepted[:,1],
                  s=10, alpha=0.3)
axes[1,1].set_title("D. ABC Posterior Distribution")
axes[1,1].set_xlabel(r"$\alpha$")
axes[1,1].set_ylabel(r"$\beta$")

plt.tight_layout()
plt.savefig("adaptive_trait_evolution_figure.png", bbox_inches="tight")
plt.show()

# ============================================================
# 8. SUMMARY OUTPUT
# ============================================================

print("Analysis completed.")
print(f"Accepted posterior samples: {len(accepted)}")
print("Figures saved at publication resolution (600 dpi).")
```

```python
# ============================================================
# 4. GAUSSIAN PROCESS REGRESSION (Publication-Level)
# ============================================================

import matplotlib.pyplot as plt
import numpy as np
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    RBF,
    ConstantKernel,
    WhiteKernel
)

# ------------------------------------------------------------
# 4.1 Model Specification
# ------------------------------------------------------------

# Kernel: signal variance × RBF + observational noise
kernel = (
    ConstantKernel(1.0, (1e-3, 1e3))
    * RBF(length_scale=5.0, length_scale_bounds=(1e-2, 1e2))
    + WhiteKernel(noise_level=0.1, noise_level_bounds=(1e-5, 1e1))
)

gp_model = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=10,
    normalize_y=True,
    random_state=42
)

# Training data
X_train = carnivore.log_body_mass.values.reshape(-1, 1)
y_train = carnivore.niche_breadth.values

# Model fitting
gp_model.fit(X_train, y_train)

# ------------------------------------------------------------
# 4.2 Posterior Prediction
# ------------------------------------------------------------

X_pred = np.linspace(
    X_train.min(),
    X_train.max(),
    300
).reshape(-1, 1)

y_mean, y_std = gp_model.predict(X_pred, return_std=True)

# 95% credible interval
ci_upper = y_mean + 1.96 * y_std
ci_lower = y_mean - 1.96 * y_std

# ------------------------------------------------------------
# 4.3 Publication-Quality Visualization
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

# Raw observations
plt.scatter(
    X_train,
    y_train,
    s=18,
    alpha=0.35,
    label="Observed Data"
)

# Posterior mean
plt.plot(
    X_pred,
    y_mean,
    linewidth=2.2,
    label="Posterior Mean"
)

# Credible interval
plt.fill_between(
    X_pred.flatten(),
    ci_lower,
    ci_upper,
    alpha=0.2,
    label="95% Credible Interval"
)

plt.xlabel("Log Body Mass")
plt.ylabel("Niche Breadth")
plt.title("Gaussian Process Inference of Trait–Niche Landscape")

plt.legend(frameon=False)
plt.tight_layout()

plt.savefig(
    "figure_gp_trait_landscape.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 4.4 Model Diagnostics
# ------------------------------------------------------------

print("Optimized Kernel:")
print(gp_model.kernel_)
print(f"Log Marginal Likelihood: {gp_model.log_marginal_likelihood(gp_model.kernel_.theta):.3f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess

# -------------------------------------------------
# 1. Reproducibility & Publication Style
# -------------------------------------------------
np.random.seed(42)

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 12,
    "figure.dpi": 300,
    "savefig.dpi": 600,
    "axes.spines.top": False,
    "axes.spines.right": False
})

# -------------------------------------------------
# 2. Simulated Ecological Data
# -------------------------------------------------
def generate_nekton():
    x = np.random.uniform(4, 16, 150)
    y = -0.02 * (x - 10)**2 + 3 + np.random.normal(0, 0.5, len(x))
    return pd.DataFrame({"log_fecundity": x, "log_longevity": y})

def generate_carnivore():
    x = np.random.uniform(4, 12, 150)
    y = -0.05 * (x - 8)**2 + 1.5 + np.random.normal(0, 0.2, len(x))
    return pd.DataFrame({"log_body_mass": x, "niche_breadth": y})

nekton = generate_nekton()
carnivore = generate_carnivore()

# -------------------------------------------------
# 3. LOESS + Bootstrap Confidence Bands + %Gap
# -------------------------------------------------
def loess_with_ci(x, y, frac=0.3, n_boot=1000):
    grid = np.linspace(x.min(), x.max(), 300)

    # Fit LOESS
    fitted = lowess(y, x, frac=frac, return_sorted=True)
    y_smooth = np.interp(grid, fitted[:,0], fitted[:,1])

    # Bootstrap
    boot_curves = []
    for _ in range(n_boot):
        idx = np.random.choice(len(x), len(x), replace=True)
        x_boot = x[idx]
        y_boot = y[idx]
        fitted_boot = lowess(y_boot, x_boot, frac=frac, return_sorted=True)
        y_boot_interp = np.interp(grid, fitted_boot[:,0], fitted_boot[:,1])
        boot_curves.append(y_boot_interp)

    boot_curves = np.array(boot_curves)
    lower = np.percentile(boot_curves, 2.5, axis=0)
    upper = np.percentile(boot_curves, 97.5, axis=0)

    # % Gap
    pct_gap = 100 * (upper - lower) / np.maximum(np.abs(y_smooth), 1e-6)

    return grid, y_smooth, lower, upper, pct_gap

# Carnivore
grid_c, smooth_c, lower_c, upper_c, gap_c = loess_with_ci(
    carnivore.log_body_mass.values,
    carnivore.niche_breadth.values
)

# Nekton
grid_n, smooth_n, lower_n, upper_n, gap_n = loess_with_ci(
    nekton.log_fecundity.values,
    nekton.log_longevity.values
)

# -------------------------------------------------
# 4. Highlight High-Impact Points (>90th percentile)
# -------------------------------------------------
def high_points(df, x_col, y_col):
    x_thresh = np.percentile(df[x_col], 90)
    y_thresh = np.percentile(df[y_col], 90)
    return df[(df[x_col] > x_thresh) & (df[y_col] > y_thresh)]

carnivore_hp = high_points(carnivore, "log_body_mass", "niche_breadth")
nekton_hp = high_points(nekton, "log_fecundity", "log_longevity")

# -------------------------------------------------
# 5. Multi-Panel Figure with Impact Colors
# -------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# Color palette
scatter_color = "#1f77b4"       # blue
ci_color = "#aec7e8"            # light blue
highlight_color = "#ff7f0e"     # orange
smooth_color = "#2ca02c"        # green

# ---- Carnivore Panel ----
axes[0].scatter(
    carnivore.log_body_mass,
    carnivore.niche_breadth,
    s=18,
    alpha=0.35,
    color=scatter_color
)
axes[0].scatter(
    carnivore_hp.log_body_mass,
    carnivore_hp.niche_breadth,
    s=40,
    alpha=0.9,
    color=highlight_color,
    label="Top 10% values"
)
axes[0].plot(grid_c, smooth_c, linewidth=2.2, color=smooth_color)
axes[0].fill_between(grid_c, lower_c, upper_c, alpha=0.25, color=ci_color)

axes[0].set_xlabel("Log Body Mass")
axes[0].set_ylabel("Niche Breadth")
axes[0].set_title("A. Carnivore Trait–Niche Landscape")
axes[0].legend()

# ---- Nekton Panel ----
axes[1].scatter(
    nekton.log_fecundity,
    nekton.log_longevity,
    s=18,
    alpha=0.35,
    color=scatter_color
)
axes[1].scatter(
    nekton_hp.log_fecundity,
    nekton_hp.log_longevity,
    s=40,
    alpha=0.9,
    color=highlight_color,
    label="Top 10% values"
)
axes[1].plot(grid_n, smooth_n, linewidth=2.2, color=smooth_color)
axes[1].fill_between(grid_n, lower_n, upper_n, alpha=0.25, color=ci_color)

axes[1].set_xlabel("Log Fecundity")
axes[1].set_ylabel("Log Longevity")
axes[1].set_title("B. Nekton Life-History Trade-off")
axes[1].legend()

plt.tight_layout()
plt.savefig(
    "nonparametric_trait_landscape_updated.pdf",
    bbox_inches="tight"
)
plt.show()

# -------------------------------------------------
# 6. Figure Caption
# -------------------------------------------------
figure_caption = """
Figure X. Nonparametric visualization of trait–fitness landscapes in simulated ecological datasets.
(A) Carnivore trait–niche landscape showing log body mass vs. niche breadth. LOESS smoothing
(green line) with 95% bootstrap confidence intervals (light blue envelope) illustrates the
nonparametric relationship. High-impact points (orange) highlight species with both large
body mass and broad niche breadth (top 10% quantiles). Percentage gap between CI and LOESS
curve (not shown) quantifies relative uncertainty along the trait gradient.

(B) Nekton life-history trade-off between log fecundity and log longevity, plotted similarly.
High-impact points emphasize species with extreme fecundity and longevity. This approach
provides a publication-quality depiction of nonlinear ecological trait–fitness relationships
without assuming parametric forms.
"""
print(figure_caption)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.nonparametric.smoothers_lowess import lowess
from scipy.stats import gaussian_kde

# -------------------------------------------------
# 1. Reproducibility
# -------------------------------------------------
np.random.seed(42)
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 12,
    "figure.dpi": 300,
    "savefig.dpi": 600,
    "axes.spines.top": False,
    "axes.spines.right": False
})

# -------------------------------------------------
# 2. Simulated Ecological Data
# -------------------------------------------------
def generate_nekton():
    x = np.random.uniform(4, 16, 150)
    y = -0.02 * (x - 10)**2 + 3 + np.random.normal(0, 0.5, len(x))
    return pd.DataFrame({"log_fecundity": x, "log_longevity": y})

def generate_carnivore():
    x = np.random.uniform(4, 12, 150)
    y = -0.05 * (x - 8)**2 + 1.5 + np.random.normal(0, 0.2, len(x))
    return pd.DataFrame({"log_body_mass": x, "niche_breadth": y})

nekton = generate_nekton()
carnivore = generate_carnivore()

# -------------------------------------------------
# 3. LOESS + Bootstrap CI
# -------------------------------------------------
def loess_with_ci(x, y, frac=0.3, n_boot=500):
    grid = np.linspace(x.min(), x.max(), 300)
    fitted = lowess(y, x, frac=frac, return_sorted=True)
    y_smooth = np.interp(grid, fitted[:,0], fitted[:,1])

    boot_curves = []
    for _ in range(n_boot):
        idx = np.random.choice(len(x), len(x), replace=True)
        x_boot = x[idx]
        y_boot = y[idx]
        fitted_boot = lowess(y_boot, x_boot, frac=frac, return_sorted=True)
        y_boot_interp = np.interp(grid, fitted_boot[:,0], fitted_boot[:,1])
        boot_curves.append(y_boot_interp)

    boot_curves = np.array(boot_curves)
    lower = np.percentile(boot_curves, 2.5, axis=0)
    upper = np.percentile(boot_curves, 97.5, axis=0)

    pct_gap = 100 * (upper - lower) / np.maximum(np.abs(y_smooth), 1e-6)

    return grid, y_smooth, lower, upper, pct_gap

# Carnivore
grid_c, smooth_c, lower_c, upper_c, gap_c = loess_with_ci(
    carnivore.log_body_mass.values,
    carnivore.niche_breadth.values
)

# Nekton
grid_n, smooth_n, lower_n, upper_n, gap_n = loess_with_ci(
    nekton.log_fecundity.values,
    nekton.log_longevity.values
)

# -------------------------------------------------
# 4. High-Impact Points
# -------------------------------------------------
def high_points(df, x_col, y_col):
    x_thresh = np.percentile(df[x_col], 90)
    y_thresh = np.percentile(df[y_col], 90)
    return df[(df[x_col] > x_thresh) & (df[y_col] > y_thresh)]

carnivore_hp = high_points(carnivore, "log_body_mass", "niche_breadth")
nekton_hp = high_points(nekton, "log_fecundity", "log_longevity")

# -------------------------------------------------
# 5. Heatmap Function
# -------------------------------------------------
def plot_heatmap(df, x_col, y_col, title, high_points_df=None, bins=100):
    x = df[x_col].values
    y = df[y_col].values

    # KDE for smooth density
    xy = np.vstack([x, y])
    kde = gaussian_kde(xy)

    # Grid for heatmap
    x_grid = np.linspace(x.min(), x.max(), bins)
    y_grid = np.linspace(y.min(), y.max(), bins)
    X, Y = np.meshgrid(x_grid, y_grid)
    Z = kde(np.vstack([X.ravel(), Y.ravel()])).reshape(X.shape)

    plt.imshow(
        Z,
        origin='lower',
        aspect='auto',
        extent=[x.min(), x.max(), y.min(), y.max()],
        cmap='viridis'
    )

    plt.colorbar(label="Density")
    plt.xlabel(x_col.replace("_", " ").title())
    plt.ylabel(y_col.replace("_", " ").title())
    plt.title(title)

    # Overlay high-impact points
    if high_points_df is not None:
        plt.scatter(
            high_points_df[x_col],
            high_points_df[y_col],
            color='orange',
            s=40,
            label='Top 10% values'
        )
        plt.legend()

# -------------------------------------------------
# 6. Plot Multi-Panel Heatmap
# -------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

plt.sca(axes[0])
plot_heatmap(
    carnivore,
    "log_body_mass",
    "niche_breadth",
    "A. Carnivore Trait–Niche Landscape",
    carnivore_hp
)

plt.sca(axes[1])
plot_heatmap(
    nekton,
    "log_fecundity",
    "log_longevity",
    "B. Nekton Life-History Trade-off",
    nekton_hp
)

plt.tight_layout()
plt.savefig("trait_fitness_heatmap.pdf", bbox_inches="tight")
plt.show()

# -------------------------------------------------
# 7. Figure Caption
# -------------------------------------------------
figure_caption = """
Figure X. Heatmap visualization of nonparametric trait–fitness landscapes in simulated ecological datasets.
(A) Carnivore trait–niche landscape. Color intensity represents density of species along log body mass
and niche breadth axes, highlighting regions with higher occurrence. High-impact points (orange) show species
in the top 10% quantiles for both traits.

(B) Nekton life-history trade-off between log fecundity and log longevity. Heatmap colors indicate
densely occupied trait combinations. High-impact points emphasize species with extreme fecundity
and longevity values. This approach provides a clear, high-impact visualization of trait distributions
without assuming parametric forms.
"""
print(figure_caption)
```
